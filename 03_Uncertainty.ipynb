# Bootstrapping Uncertainty (Optional)

Run after processing. Whole-image resampling gives MD/FA 95% interval widths and E1/E2/E3 95% cone half-angles. Results go to `14_Uncertainty`. Use the CarDpy environment.


```python
from pathlib import Path
from cardpy.uncertainty import run_dataset
from cardpy.uncertainty.plotting import plot_maps

output_directory = Path("/full/path/to/study/CarDpy_Output")
```

For a new run, it save their processing config automatically.
For an older run, pass the original full config as `config=config`; if respiratory sorting was enabled, also supply `config["respiratory"]["crop_coordinates"]` as `[x_start, x_end, y_start, y_end]` lists, one value per slice, if a config is present in the folder it will call that.


```python
# Use 400 for analysis; use fewer iterations only checks execution (e.g., 2).
result = run_dataset(output_directory, iterations=400, n_jobs=4, seed=0)
```

```python
fig = plot_maps(result, slice_index=0)
```
